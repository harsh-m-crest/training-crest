# World Choropleth Map

```python
df = pd.read_csv('2014_World_GDP')
df.head()
```

```python
data = dict(
        type = 'choropleth',
        locations = df['CODE'],
        z = df['GDP (BILLIONS)'],
        text = df['COUNTRY'],
        colorbar = {'title' : 'GDP Billions US'},
      ) 
```

```python
layout = dict(
    title = '2014 Global GDP',
    geo = dict(
        showframe = False,
        projection = {'type':'Mercator'}
    )
)
```

```python
choromap = go.Figure(data = [data],layout = layout)
iplot(choromap)
```
