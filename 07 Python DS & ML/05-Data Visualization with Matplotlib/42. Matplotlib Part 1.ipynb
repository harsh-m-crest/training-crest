```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
%matplotlib inline
```

```python
X = np.linspace(0,5,11)
y = X ** 2
```

```python
X
```

```python
y
```

```python
# Fuctional
plt.plot(X,y)
plt.xlabel('X label')
plt.ylabel('Y Label')
plt.title('Title')
```

```python
plt.plot(X,y,'r-')
```

```python
plt.subplot(1,2,1)
plt.plot(X,y,'r')
plt.subplot(1,2,2)
plt.plot(y,X)
```

```python
# OO
fig = plt.figure()
axes = fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(X,y)
axes.set_xlabel('X Label')
axes.set_ylabel('Y label')
axes.set_title('Set Title')
```

```python
fig = plt.figure()
axes1 = fig.add_axes([0.1,0.1,0.8,0.8])
axes2 = fig.add_axes([0.2,0.5,0.4,0.3])

axes1.plot(X,y)
axes1.set_title('Larger Plot')
axes2.plot(y,X)
axes2.set_title('Smaller Plot')
```
