# Lambda Expressions Review

```python
def square(num):
    result = num**2
    return result
```

```python
square(4)
```

```python
def square(num):return num**2
```

```python
square(4)
```

```python
sq = lambda num : num**2
```

```python
sq(5)
```

```python
even = lambda num: num%2 == 0
```

```python
even(3)
```

```python
even(4)
```

```python
first = lambda s: s[0]
```

```python
first('asdfasdf')
```

```python
rev = lambda s: s[::-1]
```

```python
rev('abcde')
```

```python
adderlam = lambda x,y : x+y
```

```python
adderlam(2,3)
```

```python
adderlam(3,3)
```
