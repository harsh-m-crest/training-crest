```python
class Dog():

    species = 'Mamal'
    def __init__(self,breed,name ):
        self.breed = breed
        self.name = name

    # operations/Actons ----> Methods
    def bark(self,number):
        print('WOFH! My name is {} and the number is {}'.format(self.name,number))
```

```python
my_dog = Dog('Lab','Frankie')
```

```python
type(my_dog)
```

```python
my_dog.bark(10)
```

```python
class Circle():


    # CLASS OBJECT ATTRIBUTE
    pi = 3.14

    def __init__(self, radius=1):
        self.radius = radius
        self.area = radius * radius * Circle.pi

    # METHOD
    def get_circumference(self):
        return self.radius * Circle.pi * 2
```

```python
my_circle = Circle(30)
```

```python
my_circle.pi
```

```python
my_circle.radius
```

```python
my_circle.area
```

```python
my_circle.get_circumference()
```
