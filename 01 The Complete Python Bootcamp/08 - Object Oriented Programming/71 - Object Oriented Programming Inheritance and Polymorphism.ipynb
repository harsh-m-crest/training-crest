```python
class Animal():
    def __init__(self):
        print('Animal created')

    def who_am_i(self):
        print('I am an animal')

    def eat(self):
        print("I am eating")
```

```python
myanimal =Animal()
```

```python
myanimal.eat()
```

```python
myanimal.who_am_i()
```

```python
class Dog(Animal):
    def __init__(self):
        Animal.__init__(self)
        print("Dog Created")

    def who_am_i(self):
        print('I am a dog!')

    def bark(self):
        print('WOOF!')
    
```

```python
mydog = Dog()
```

```python
 mydog.who_am_i()
```

```python
mydog.bark()
```

```python
class Dog():

    def __init__(self,name):
        self.name = name

    def speak(self):
        return self.name + ' says woof!'
```

```python
class Cat():
    def __init__(self,name):
        self.name = name

    def speak(self):
        return self.name + " says meow!"
```

```python
niko = Dog('niko')
felix = Cat('felix')
```

```python
print(niko.speak())
```

```python
print(felix.speak())
```

```python
for pet in [niko,felix]:
    print (type(pet))
    print(type(pet.speak()))
```

```python
def pet_speak(pet):
    print(pet.speak())
```

```python
pet_speak(niko)
```

```python
pet_speak(niko)
```
