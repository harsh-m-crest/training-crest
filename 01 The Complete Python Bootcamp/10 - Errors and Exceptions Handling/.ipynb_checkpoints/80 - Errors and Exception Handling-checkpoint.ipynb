```python
try:
    f = open('testfile','w')
    f.write('Write a test line')
except:
    print("All other exceptions!")
finally:
    print("I always run")
```

```python
try:
    result = int(input("Please provide a number: "))
except:
    print("Whoops! That is not a number")
finally:
    print("End of try/except/finally")
```

```python
def ask_for_int():
    while True:
        try:
            result = int(input("Please provide a number: "))
        except:
            print("Whoops! That is not a number")
            continue
        else:
            print("Yes Thank you")
            break
        finally:
            print("End of try/except/finally")
            print('I will always run at the end!')
```

```python
ask_for_int()
```
