```python
my_nums =[1,2,3,4,5]
```

```python
list(map(square,my_nums))
```

```python
def splicer(mystring):
    if len(mystring)%2 == 0:
        return "EVEN"
    else:
        return mystring[0]
```

```python
names = ['Andy','Eve','Sally']
```

```python
list(map(splicer,names))
```

```python
def check_even(num):
    return num % 2 == 0
```

```python
mynums = [1,2,3,4,5,6]
```

```python
list(filter(check_even,mynums))
```

```python
for n in filter(check_even,mynums):
    print(n)
```
