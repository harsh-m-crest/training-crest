# Ex -1: Integer to string conversion in any base using recussrion

```python
def int_to_base(n,base):
    digits = '0123456789ABCDEF'

    if n < base:
        return digits[n]
    else:
        return int_to_base(n // base, base) + digits[n % base]

print(int_to_base(2835,16))
```

```python
int_to_base(100,10)
```

# Ex -2: Sum of digits of an integer Using Recusion

```python
def sum_of_digits(n):
    if n == 0:
        return 0
    return n%10 + sum_of_digits(n//10)

digits = 143
sum_of_digits(digits)
```
