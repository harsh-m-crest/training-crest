# Ex -1: Vowel Between Consonants Substring

```python
def vowel_between_consonanats(s):
    vowels = ['a','e','i','o','u']
    for i in range(len(s)-1):
        if s[i] in vowels and s[i-1] not in vowels and s[i+1] not in vowels:
            return s[i-1: i+2]
    return -1
s = 'string'
vowel_between_consonanats(s)
```

# Ex -2: strings with the most unique characters

```python
# Input
l = ['cat', 'catatatatctsa', 'abcdefhijklmnop', '124259239185125', '', 'foo', 'unique']

def most_unique(l):
    max_unique_count = 0
    result = ""

    for word in l:
        uniquechars = set(word)
        uniquecount = len(uniquechars)

        if uniquecount > max_unique_count:
            max_unique_count = uniquecount
            result = word

    return result

print(most_unique(l))
```
