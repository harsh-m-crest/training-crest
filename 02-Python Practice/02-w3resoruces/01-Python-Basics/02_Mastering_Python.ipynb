# Ex - 2: Replace all even numbers in a list with the next odd number:

```python
def replace_eve(l):
    for i in range(len(l)):
        if l[i]% 2 == 0:
            l[i] = l[i]+1
    return l

l = [2, 5, 8, 12, 15]
replace_eve(l)
```
