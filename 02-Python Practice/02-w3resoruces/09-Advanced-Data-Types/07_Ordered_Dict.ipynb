# OrderedDict from Random Data

```python
import random
from collections import OrderedDict
```

```python
def random_ascii():
    return chr(random.randint(65,90))

ordered_dict = OrderedDict()

for i in range(10):
    key = random_ascii()
    value = random.randint(1,50)
    ordered_dict[key] = value
print(ordered_dict)
```
