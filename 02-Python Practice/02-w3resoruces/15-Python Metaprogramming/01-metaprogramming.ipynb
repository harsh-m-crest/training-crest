# Python function "genreate_function" that takes a function name and a body as strings, and returns a dynamically genrated function

```python
def generate_functions(name,body):
    local_namespace = {}
    exec(f"def {name}():\n    {body}", {}, local_namespace)
    return local_namespace[name]
```

```python
function_name = 'dynamic_function'
function_body = "print('Hello from the dynamically genreated function!')"

dynamic_func = generate_functions(function_name, function_body)

dynamic_func()
```
