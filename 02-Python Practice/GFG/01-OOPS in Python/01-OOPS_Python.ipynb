# Inheritance

```python
class Dog:
    def __init__(self,name):
        self.name = name

        def display_name(self):
            print(f"Dog's Name:{self.name}")
```

```python
class Labrador(Dog):
    def sound(self):
        print("Labrador woofs")
```

```python
class GuideDog(Labrador):
    def guide(self):
        print(f"{self.name}Guides the way!")
```

# Polymorphism

```python
class Calculator:
    def add(self, *args):
        return sum(args)
```

```python
calc = Calculator()
print(calc.add(5,10))
print(calc.add(5,10,15))
print(calc.add(1,2,3,4))
```

```python
class Animal:
    def speak(self):
        return "I am an animal."

class Dog(Animal):
    def speak(self):
        return "Woof !"
```

```python
Dog().speak()
```

```python
class Cat:
    def speak(self):
        return "Meow !"
```

```python
def make_animal_speak(animal):
    return animal.speak()
```

```python
print(make_animal_speak(Cat()))
print(make_animal_speak(Dog()))
```

# Encapsulation

```python
class Dog:
    def __init__(self,name,breed,age):
        self.name = name
        self._breed = breed
        self.__age = age
    
```

```python
def get_info(self):
    return f"Name : {self.name}, Breed: {self._breed}, Age: {self.__age}"
```
